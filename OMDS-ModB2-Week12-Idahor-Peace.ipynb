# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np

categories = ['Product A', 'Product B', 'Product C', 'Product D', 'Product E']
years = [2019, 2020, 2021, 2022, 2023, 2024, 2025]

data = {
    'Product A': [390, 415, 390, 310, 265, 250, None],
    'Product B': [360, 405, 375, 280, 245, 240, None],
    'Product C': [None, None, None, 95,  175, 180, None],
    'Product D': [None, None, None, 150, 255, 215, None],
    'Product E': [None, None, None, None, None, 95,  210],
}

colors = ['#d9d9d9', '#bfbfbf', '#a6a6a6', '#808080', '#595959', '#333333', '#000000']

x = np.arange(len(categories))
n_years = len(years)
bar_width = 0.11
offsets = np.linspace(-(n_years - 1) / 2, (n_years - 1) / 2, n_years) * bar_width

fig, ax = plt.subplots(figsize=(10, 6))
fig.patch.set_facecolor('white')
ax.set_facecolor('white')

for i, (year, color) in enumerate(zip(years, colors)):
    values = [data[cat][i] if data[cat][i] is not None else 0 for cat in categories]
    bars = ax.bar(x + offsets[i], values, width=bar_width, color=color, label=str(year))
    # Zero-value bars should be invisible
    for bar, val in zip(bars, [data[cat][i] for cat in categories]):
        if val is None:
            bar.set_visible(False)

ax.set_title('Average Retail Product Price Per Year', fontsize=13, fontweight='normal', pad=12)
ax.set_xticks(x)
ax.set_xticklabels(categories, fontsize=11)
ax.set_ylim(0, 500)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f'${int(v)}'))
ax.tick_params(axis='both', length=0)
ax.set_yticks([0, 100, 200, 300, 400, 500])

# Grid lines only horizontal
ax.yaxis.grid(True, color='#e0e0e0', linewidth=0.8)
ax.set_axisbelow(True)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_color('#cccccc')

# Legend below chart
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles, labels, ncol=len(years), loc='upper center',
          bbox_to_anchor=(0.5, -0.08), frameon=False,
          fontsize=10, handlelength=1.2, handleheight=0.8)

plt.tight_layout()
plt.savefig('retail_price_chart.png', dpi=150, bbox_inches='tight')
plt.show()
```
