# 2. Weekly graph question

Suppose you wanted to show the nitrate and phosphate level in a water sample.  What are the advantages and disadvantages of showing this as an area plot, as opposed to two separate line graphs?  How would you adjust the graphs shown to improve the presentation?

```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series_2 =  100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series_2)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

```python
df = pd.DataFrame({'ser1': time_series, 'ser2': time_series_2}, index = datearray)
df.plot.area()
```

### Advantages of Area Plot
* It is a great way to show how the two quantities compare against each other. It is visually easy to see as long as bold colors are sued to show the distictions between the two quantities.

* If the area plot is stacked it is easier to see how much of the total sum each quantity contributes. 
### Disadvantages of Area Plot 
* There could be some overlapping. If you are comparing multiple quantities, it can become a mess of overlapping information that is hard to decipher.  

* If in a stacked area plot there is a spike in one of the quantities, the spike is represented visually for all the quantities. 

### Improving the graphs. 
* For the lines graphs I would plot both quantities on the same graph. They are on the same time scale so it allows me to see better if there is a correlaton between the two quantities. I would then make each like distinct to what each line represents and add a legend. 

* For the area plot, I will make the legend more detailed, as to indicate which color is which quantity. I will also populate the x and the y axis as well.

```python
import kagglehub
import os
# Download latest version
path = kagglehub.dataset_download("ammaraahmad/us-ecommerce-record-2020")

ecom_data = pd.read_csv(os.path.join(path, os.listdir(path)[0]), encoding='latin1')
ecom_data.head()
```

```python
ecom_data.info()
ecom_data.describe()
```

```python
ecom_data_numeric = ecom_data.select_dtypes(include='number').drop(columns=['Row ID','Postal Code'])

ecom_data_numeric.columns
```

```python
ecom_data_corr_matrix = ecom_data_numeric.corr()
ecom_data_corr_matrix
```

```python
# Creating heatmap of correlation matrix
import seaborn as sns
import matplotlib.pyplot as plt

# 3. Create the heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(ecom_data_corr_matrix, 
            annot=True,          # Shows the numeric correlation value in each cell
            cmap='coolwarm',     # Red for positive, blue for negative correlation
            fmt=".2f",           # Limits decimals to 2 places
            linewidths=0.5,      # Adds spacing between squares
            center=0)            # Sets the color balance at 0

plt.title('Correlation Matrix: US E-commerce Records 2020', fontsize=16)
plt.show()
```

```python
# Creating scatter plots of useful correlation 
threshold = 0.5
corr_pairs = []

# We iterate through the columns and only look at the upper triangle of the matrix
# to avoid getting (A, B) and (B, A) as separate pairs.
for i in range(len(ecom_data_corr_matrix.columns)):
    for j in range(i + 1, len(ecom_data_corr_matrix.columns)):
        col1 = ecom_data_corr_matrix.columns[i]
        col2 = ecom_data_corr_matrix.columns[j]
        correlation = ecom_data_corr_matrix.iloc[i, j]
        
        if correlation > threshold:
            corr_pairs.append((col1, col2, correlation))

# 2. Generate scatter plots for each qualifying pair
num_pairs = len(corr_pairs)

if num_pairs > 0:
    # Set up a grid for multiple plots (1 row, many columns)
    fig, axes = plt.subplots(nrows=1, ncols=num_pairs, figsize=(5 * num_pairs, 5))
    
    # If there's only one pair, axes is not a list by default
    if num_pairs == 1:
        axes = [axes]
    
    for idx, (col1, col2, corr) in enumerate(corr_pairs):
        sns.scatterplot(data=ecom_data_numeric, x=col1, y=col2, ax=axes[idx])
        axes[idx].set_title(f'{col1} vs {col2}\nCorrelation: {corr:.2f}')
    
    plt.tight_layout()
    plt.show()
else:
    print("No pairs found with a correlation greater than 0.5.")
```

```python
# 1. Identify all columns that participate in a correlation > 0.5
threshold = 0.5
high_corr_cols = set()

for i in range(len(ecom_data_corr_matrix.columns)):
    for j in range(i + 1, len(ecom_data_corr_matrix.columns)):
        if abs(ecom_data_corr_matrix.iloc[i, j]) > threshold:
            high_corr_cols.add(ecom_data_corr_matrix.columns[i])
            high_corr_cols.add(ecom_data_corr_matrix.columns[j])

# 2. Subset the numeric data to only include these columns
# We convert the set back to a list
cols_to_plot = list(high_corr_cols)

if len(cols_to_plot) > 1:
    # 3. Create the pair plot
    # kind='reg' adds a regression line to the scatter plots
    # diag_kind='kde' shows the density distribution on the diagonal
    pp = sns.pairplot(ecom_data_numeric[cols_to_plot], 
                      kind='reg', 
                      diag_kind='kde',
                      plot_kws={'line_kws':{'color':'red'}, 'scatter_kws': {'alpha': 0.5}})
    
    pp.fig.suptitle(f'Pair Plot of Variables with Correlation > {threshold}', y=1.02, fontsize=16)
    plt.show()
else:
    print(f"No columns found with a correlation above {threshold}.")
```
