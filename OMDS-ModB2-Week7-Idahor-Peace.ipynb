# 2. Weekly graph question

Below are the graphs from last week's weekly graph question.  Suppose you wanted to cause the viewer to examine the last three months of data first.  How could you use preattentive attributes to do that?  Suggest several possibilities.

```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

### Making the Last Three Months Pop in a Time Series Plot

### 1. Color (Hue and Intensity)
I find color to be my most powerful tool for guiding the viewer’s eye. I can visually "mute" the first nine months and make the last three months really stand out.

**My Strategy:** I plot the first nine months in a neutral gray and the last three months in a bold, high-contrast color like bright red or orange.


### 2. Enclosure
I know that drawing a boundary around a set of points helps the brain group them and treat them as important.

**My Strategy:** I place a shaded box (a "highlight zone") behind the data points for October, November, and December.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
sns.set(style="whitegrid")

# Example data (you can replace with your actual data)
data = {
    'Quarter': ['Q1 2023','Q2 2023','Q3 2023','Q4 2023',
                'Q1 2024','Q2 2024','Q3 2024','Q4 2024',
                'Q1 2025','Q2 2025','Q3 2025'],
    'Exceed': [30, 28, 25, 27, 32, 30, 28, 29, 25, 20, 18],
    'Meet': [50, 52, 55, 53, 55, 53, 55, 54, 45, 47, 40],
    'Miss': [20, 20, 20, 20, 13, 17, 17, 17, 30, 33, 42]
}

df = pd.DataFrame(data)

# Plot stacked bars
fig, ax = plt.subplots(figsize=(10,6))

ax.bar(df['Quarter'], df['Exceed'], label='Exceed', color='lightgray')
ax.bar(df['Quarter'], df['Meet'], bottom=df['Exceed'], label='Meet', color='gray')
ax.bar(df['Quarter'], df['Miss'], bottom=df['Exceed']+df['Meet'], label='Miss', color='firebrick')

# Add percentages on top of the bars for the last bar as in your example
for i, row in df.iterrows():
    if row['Miss'] > 0:
        ax.text(i, row['Exceed'] + row['Meet'] + row['Miss']/2, f"{row['Miss']}%", ha='center', va='center', color='white', fontweight='bold')

# Titles and labels
ax.set_ylabel('% of total projects')
ax.set_title('Goal attainment over time')
ax.legend(title='Goal')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
